```python
import sys
from pathlib import Path

# Find the project root
current = Path.cwd().resolve()

for folder in [current, *current.parents]:
    if (folder / "include" / "clean").is_dir():
        PROJECT_ROOT = folder
        break
else:
    raise FileNotFoundError(
        "Cannot find include/clean. Check where you extracted the ZIP."
    )

# Add include to Python's import search path
INCLUDE_DIR = PROJECT_ROOT / "include"

if str(INCLUDE_DIR) not in sys.path:
    sys.path.insert(0, str(INCLUDE_DIR))

print("Project root:", PROJECT_ROOT)
print("Clean package:", INCLUDE_DIR / "clean")

from clean import run_all

print("Clean module imported successfully!")
```
